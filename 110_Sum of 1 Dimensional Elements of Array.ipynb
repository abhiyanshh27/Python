```python
import numpy as np
a=[]
size = int(input("How many Numbers?"))
for i in range (size):
    val=int(input("Enter number:"))
    a.append(val)
arr=np.array(a)
for i in range(arr.size):
    print(arr[i])
sum=0
for i in range (arr.size):
    sum+=arr[i]
print("Sum of elements = ",sum)
print(arr)
```
